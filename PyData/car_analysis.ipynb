```python
df['Engine_Category'] = pd.cut(df['Engine_CC'], bins=[0, 1000, 1500, 2000, 3000, 4000, 5000], labels=['Below 1.0L', '1.0L-1.5L', '1.5L-2.0L', '2.0L-3.0L', '3.0L-4.0L', '4.0L-5.0L'])
```

```python
df['Mileage_Category'] = pd.cut(
    df['Mileage'],
    bins=[0, 15, 20, 25, 30],
    labels=['Low', 'Average', 'Good', 'Excellent']
)
```

```python
brand_price = (
    df.groupby('Brand')['Price_Lakh']
      .mean()
      .sort_values(ascending=False)
)

brand_price
```

```python
import matplotlib.pyplot as plt
```

```python
brand_price.plot(kind='bar', figsize=(12, 6), color='skyblue')
plt.title('Average Price of Cars by Brand (in Lakh)')
plt.xlabel('Brand')
plt.ylabel('Average Price (Lakh)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df.columns
```

```python
import matplotlib.pyplot as plt

brand_price.plot(kind='bar', figsize=(10, 5))

plt.title('Average Car Price by Brand')
plt.xlabel('Brand')
plt.ylabel('Average Price (₹ Lakhs)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
brand_count = df['Brand'].value_counts()

brand_count.plot(kind='bar', figsize=(10,5))

plt.title('Number of Cars by Brand')
plt.xlabel('Brand')
plt.ylabel('Number of Cars')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
fuel = df['Fuel_Type'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(fuel, labels=fuel.index, autopct='%1.1f%%', startangle=90)
plt.title('Fuel Type Distribution')
plt.show()
```

```python
trans = df['Transmission'].value_counts()

trans.plot(kind='bar', figsize=(6,4))

plt.title('Manual vs Automatic')
plt.xlabel('Transmission')
plt.ylabel('Cars')
plt.show()
```

```python
year_price = df.groupby('Year')['Price_Lakh'].mean()

year_price.plot(kind='line', marker='o', figsize=(8,5))

plt.title('Average Price by Year')
plt.xlabel('Year')
plt.ylabel('Price (Lakh)')
plt.grid(True)
plt.show()
```

```python
top_models = (
    df.groupby('Model')['Price_Lakh']
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

top_models.plot(kind='barh', figsize=(8,5))

plt.title('Top 10 Expensive Models')
plt.xlabel('Price (Lakh)')
plt.show()
```

```python
service = df.groupby('Brand')['service_cost_thousand'].mean()

service.plot(kind='bar', figsize=(10,5))

plt.title('Average Service Cost by Brand')
plt.ylabel('Service Cost (Thousand ₹)')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Mileage'], df['Price_Lakh'], alpha=0.5)

plt.title('Price vs Mileage')
plt.xlabel('Mileage (km/l)')
plt.ylabel('Price (Lakh)')
plt.show()
```
